```python
from statsbombpy import sb
import pandas as pd

# Set display options to inspect large dataframes easily
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
```

```python
# Fetch available open competitions
competitions = sb.competitions()

# View the available tournaments (e.g., La Liga, FIFA World Cup, NWSL)
competitions.country_name.unique()
```

```python
international=competitions[
    competitions.competition_name.isin(
        ['UEFA Euro', 'FIFA World Cup', 'Copa America']
    )]

international=international[international.season_name.astype(int) >= 2022]
```

```python
comp_id = international.competition_id.unique().tolist()
comp_id
```

```python
season_id = international.season_id.unique().tolist()
season_id
```

```python


all_matches = []

for _, row in international[['competition_id', 'season_id']].drop_duplicates().iterrows():
    try:
        matches = sb.matches(
            competition_id=row['competition_id'],
            season_id=row['season_id']
        )
        all_matches.append(matches)
    except Exception as e:
        print(f"Error for competition {row['competition_id']} season {row['season_id']}: {e}")

all_matches = pd.concat(all_matches, ignore_index=True)
print(all_matches.columns)
all_matches=all_matches[['match_id',
             'match_date',
             'home_team',
             'away_team',
             'home_score',
             'away_score','competition','season','competition_stage' ]]
```

```python
all_matches
```

```python
# France matches
france = all_matches[
    (all_matches.home_team == 'France') |
    (all_matches.away_team == 'France')
]

# Match IDs
france_match_ids = france['match_id'].tolist()

france_match_ids[:5]
```

```python
france_events = []

for match_id in france_match_ids:

    events = sb.events(match_id=match_id)

    events['match_id'] = match_id  # ensure match_id is present

    france_events.append(events)

france_events = pd.concat(france_events, ignore_index=True)
```

```python
france_events.type.unique()
```

```python
ball_recoveries = france_events[
    france_events.type == 'Ball Recovery'
].dropna(axis=1, how='all')

ball_recoveries.head()
```

```python
pass_df = france_events[
    france_events.type == 'Pass'
].dropna(axis=1, how='all')

pass_df.head()
```

```python
from statsbombpy import sb
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mplsoccer import Pitch
```

```python
match_id = france.match_id.iloc[0]  # choose a match
events = sb.events(match_id=match_id)
```

```python
passes = events[
    (events['type'] == 'Pass') &
    (events['team'] == 'France') &
    (events['pass_outcome'].isna())  # completed passes
].copy()
```

```python
passes = passes[['player', 'pass_recipient', 'location']].dropna()

passes['x'] = passes['location'].apply(lambda x: x[0])
passes['y'] = passes['location'].apply(lambda x: x[1])
```

```python
avg_pos = (
    passes.groupby('player')
    .agg(
        x=('x', 'mean'),
        y=('y', 'mean'),
        count=('x', 'count')
    )
    .reset_index()
)
```

```python
pass_links = (
    passes.groupby(['player', 'pass_recipient'])
    .size()
    .reset_index(name='passes')
)

# keep only strong connections
pass_links = pass_links[pass_links['passes'] >= 3]
```

```python
pitch = Pitch(
    pitch_type='statsbomb',
    pitch_color='#22312b',
    line_color='white'
)

fig, ax = pitch.draw(figsize=(12, 8))

# Draw connections
for _, row in pass_links.iterrows():

    p1 = avg_pos[avg_pos.player == row['player']]
    p2 = avg_pos[avg_pos.player == row['pass_recipient']]

    if len(p1) == 0 or len(p2) == 0:
        continue

    x1, y1 = p1[['x', 'y']].iloc[0]
    x2, y2 = p2[['x', 'y']].iloc[0]

    pitch.lines(
        x1, y1, x2, y2,
        lw=row['passes'] / 2,
        color='white',
        alpha=0.5,
        ax=ax
    )

# Draw players
pitch.scatter(
    avg_pos.x,
    avg_pos.y,
    s=avg_pos['count'] * 5,
    color='red',
    edgecolors='white',
    linewidth=2,
    ax=ax,
    zorder=3
)

# Labels
for _, row in avg_pos.iterrows():
    ax.text(
        row.x,
        row.y,
        row.player.split()[-1],
        ha='center',
        va='center',
        color='white',
        fontsize=8,
        zorder=4
    )

plt.title('France Passing Network', color='white', fontsize=18)
plt.show()
```
